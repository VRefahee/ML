```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x=np.array([1,2,3,4]).reshape(-1,1)
y=np.array([1,3,5,7]).reshape(-1,1)

plt.plot(x,y,'o',color='red')
plt.show()
```

#####   ------------------------رگرسیون تک متغیره و خطی---------------------------

```python
from sklearn.linear_model import LinearRegression

model=LinearRegression()
model.fit(x,y)
```

```python
#   y= m * x + c

m = model.coef_[0]     # ضریب x
c = model.intercept_
print(m,c)

```

```python
plt.plot(x,y,'o',color='red')
plt.plot(x,model.predict(x))
plt.show()
```

```python
result=model.predict([[5]])

result
```

### -------------------------رگرسیون غیر خطی--------------------

```python
years=np.array([0,1,2,3,4,6,8,10]).reshape(-1,1)
salaris=np.array([30000,40000,55000,60000,70000,80000,85000,87000]).reshape(-1,1)
```

```python
plt.plot(years,salaris,'*')
plt.show()
```

```python
from sklearn.tree import DecisionTreeRegressor

model=DecisionTreeRegressor(max_depth=2)
model.fit(years,salaris)

y_predict=model.predict(years)

plt.plot(years,salaris,'o',color='green')
plt.plot(years,y_predict,'c',color='red')

plt.show()
```
