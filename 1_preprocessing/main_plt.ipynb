```python
import matplotlib.pyplot as plt
```

```python
x=[1,2,3,4,5,6,7,8,9]
y=[1,4,7,6,8,4,5,7,10]
plt.plot(x,y,marker="o",color='red')
plt.title('example')
plt.xlabel('x')
plt.ylabel('y')
plt.show()

```

```python
x=["red","green","blue","orange"]
y=[10,5,9,3]


plt.bar(x,y)
plt.title('example')
plt.xlabel('color')
plt.ylabel('count')
plt.show()
```

```python
x=["red","green","blue","orange"]
y=[10,5,9,3]

plt.pie(y,labels=x , autopct='%1.1f%%' ,colors=x)
plt.title('test')
plt.show()
```

```python
import numpy as np
x=np.random.rand(1000,1)

plt.hist(x ,bins=10,)
plt.show()
```
