```python
y_true=[1,1,0,0,1,0,0,0,1,1]
y_pred=[1,1,0,0,1,0,0,0,1,1]
```

###  ارزیابی داده ها

##### accuracy_score         درصد درست بودن کل جواب ها
##### precision_score      درصد مثبت بودن جواب های حدس زده نسبت به جواب واقعی  
##### recall_score            درصد درست بودن جواب واقعی نسبت به جواب حدس زده



```python
from sklearn.metrics import accuracy_score

acc=accuracy_score(y_true,y_pred)
acc
```

```python
from sklearn.metrics import precision_score

pr=precision_score(y_true,y_pred)
pr
```

```python
from sklearn.metrics import recall_score

r=recall_score(y_true,y_pred)
r
```
